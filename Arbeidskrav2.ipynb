<h1>Oppgave 5</h1>
<!--
    Lag et program med en funksjon som tar a og b som inn-argumenter og
    som så regner ut arealet og «ytre» omkrets til en figur satt sammen av en rettvinklet trekant og en halvsirkel,
    se figuren under. Med «ytre» omkrets menes samlet lengde av de sorte strekene. 
    Funksjonen skal returnere arealet og «ytre» omkrets, som så skrives til skjerm med passende tekst.
-->

```python
import numpy as np
import math as math #Jeg fant det på w3schools

def figur(a, b):
    r = a/2  #Diameter er a, så må dele på to for å få radius
    c = math.sqrt(a**2 + b**2) #pythagoras og kvadrat fra w3school: math.sqrt(x)
    areal = (np.pi*(r**2)/2) + ((a*b)/2) #Jeg er usikker på om beregningen er riktig
    omkrets = ((2 * np.pi * r)/2) + (b+c) #Jeg er usikker på beregningen er riktig
    return (areal, omkrets)

# Spørre brukeren om a og b:
a_tall = float(input("Skriv tallet for a:"))
b_tall = float(input("Skriv tallet for b:"))

figur (a_tall, b_tall) #Bruke funksjonen på brukerens tall

print("Arealet er:", areal)
print("Ytre omkrets er:", omkrets)
```
